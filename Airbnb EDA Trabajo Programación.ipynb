## Comenzamos a analizar las relaciones entre variables

```python
Room_columns = ['Price', 'Shared Room', 'Private Room', 'Person Capacity', 'Superhost', 'Multiple Rooms', 'Business', 'Cleanliness Rating', 'Guest Satisfaction', 'Bedrooms', 'City Center (km)', 'Metro Distance (km)']
correlacion_room = df[Room_columns].corr()


fig = px.imshow(correlacion_room,
                text_auto='.3f',
                title="Correlación Price y Shared Room, Private Room, Person Capacity",
                labels=dict(color="Correlación"),
                height=800, width=800)
fig.update_layout(template="plotly_white")
fig.show()
```

Al crear la matriz de correlación con todas las variables, excepto las que eran objeto, notamos algunas relaciones interesantes como la relación negativa con Price de algunas variables que podríamos pensar que aumentan el precio. Esto se debe a que algunas de estas variables presentan un alto grado de multicolinealidad y por eso nos dan resultados distintos de los reales como ya veremos luego. Otra variable importante es guest satisfaction que tiene varias correlaciones, y Superhost al ser un término que ha introducido Airbnb por lo que es interesante estudiar el impacto de esta variable sobre el resto.

### Precio vs Roomtype, Shared Room, Private Room y Person Capacity

```python
df[['Room Type', 'Shared Room', 'Private Room', 'Person Capacity']].info()
```

Tenemos 1 variable tipo objeto porque contiene tecto, 2 variables categóricas dicotómicas y una numérica discreta (Person capacity) con pocos valores por lo que podemos utilizar gráficos de tarta para estudiar la frecuencia en dichas variables

```python
Room_columns = ['Room Type', 'Shared Room', 'Private Room', 'Person Capacity']

for col in Room_columns:
    count_data = df[col].value_counts().reset_index()

    fig = px.pie(count_data,
                 names=col, values='count',
                 title=f'Frecuencia de categorías en {col}',
                 labels={col: col, 'count': 'Frecuencia'})
    fig.update_layout(template="plotly_white")
    fig.show()
```

Viendo las frecuencias de estas variables nos damos cuenta de que las variables Room Type y Shared Room están altamente correlacionadas. "Entire home/apt" y "Private room" en Room Type equivalen a "false" en Shared Room.


Para estudiar cual de estas variables tiene un mayor impacto en "Price" utilizo una matriz de correlación y como sabemos que Room type y Shared Room son prácticamente la misma variable nos quedamos con Shared Room porque es tipo float y se puede usar para la matriz de correlación.

```python
Room_columns = ['Price', 'Shared Room', 'Private Room', 'Person Capacity']
correlacion_room = df[Room_columns].corr()


fig = px.imshow(correlacion_room,
                text_auto=True,
                title="Correlación Price y Shared Room, Private Room, Person Capacity",
                labels=dict(color="Correlación"))
fig.update_layout(template="plotly_white")
fig.show()
```

Para estudiar la matriz de correlación debemos saber que:
- Valores cerca de 1 → correlación positiva fuerte
- Valores cerca de -1 → correlación negativa fuerte
- Valores cerca de 0 → poca o ninguna relación

Teniendo esto en cuenta las correlaciones significativas que encontramos en la matriz son:
- Price y Person capacity 0.18: implica que la capacidad influye en el precio pero de forma débil. A más capacidad mayor precio.
- Price y Private Room -0.15: aunque parezca contradictorio que disponer de un cuarto privado baje el precio si miramos a la base de datos nos damos cuenta de que en la columna Room Type en los cuartos privados Private Room es "True" mientras que si es tipo de cuarto es Entire home/apt es "False" y el alquilar un apartamento entero es más caro que alquilar un cuarto privado, por lo que esta relación explica por qué Private Room baja el Precio aunque no tenga sentido.
- Private Room y Person Capacity -0.47: teniendo en cuenta la explicación anterior tiene sentido que el hecho de que el cuarto sea privado baje la capacidad de personas.


### Precio vs City Center (km), Metro Distance (km)

```python
Distance_columns = ['City Center (km)', 'Metro Distance (km)']
for col in Distance_columns:
    fig = px.histogram(df, x=col, nbins=20, title=f"Distribución de {col}")
    fig.update_layout(xaxis_title=col, yaxis_title="Frecuencia", template="plotly_white")
    fig.show()
```

Como podemos ver la mayoría de estancias se encuentran a menos de 5km de la capital y menos de 1.5km del metro

```python
Distance_columns = ['City Center (km)', 'Metro Distance (km)']
for col in Distance_columns:
    # Limito el precio a 5000 porque los atípicos rompían la escala de color y dejaban la mayoría de los datos en azul.
    fig = px.scatter(df, x=col, y='Price', color='Price', range_color=[0, 5000], trendline='ols', title=f"Relación Price con {col}")
    fig.update_layout(xaxis_title=col, yaxis_title="Precio", template="plotly_white")
    fig.show()
```

Utilizando este gráfico scatterplot podemos ver que existe una relación directa entre cercanía al metro y al centro de la ciudad con el precio gracias a la línea de tendencia.
También nos damos cuenta de que la cercanía al metro aumenta más el precio que la distancia al centro de la ciudad, en concreto, los precios más elevados se encuentran a menos de 500 metros del metro mientras que los precios a menos de 5km de el centro se mantienen relativamente estables.


## Relaciones con la Satisfacción

### Guest satisfaction vs Cleanliness

Vemos que no existe mucha dispersión por lo que la relación es fuerte y la pendiente de la línea de regresión es positiva lo que nos indica que la limpieza influye fuertemente en la satisfacción de los huespedes.

### Guest Satisfaction vs Price

```python
df['Price_bin'] = pd.cut(df['Price'],bins=[0, 200, 400, 800, df['Price'].max()],
                                    labels=['0–200', '200–400', '400–800', '+800'])

px.violin(df, x='Price_bin', y='Guest Satisfaction', title='Distribución Guest Satisfaction en Price', template='plotly_white', color='Price_bin')
```

Debido a que Price es una variable numérica contínua existen muchos precios distintos por lo que para poder representarlos creo 4 grupos de precios: '0–200', '200–400', '400–800' y '+800'.
El violin plot nos permite ver como los ratings de  satisfaccion de los clientes se encuentran en más de 80 independientemente del grupo de precios en el que se encuentre el airbnb por lo que podemos deducir que el precio no afecta a la satisfacción del huesped.

### Guest Satisfaction vs Price, Cleanliness y Superhost

```python
Satisfaction_columns = ['Guest Satisfaction','Price', 'Cleanliness Rating', 'Superhost']
correlacion_room = df[Satisfaction_columns].corr()


fig = px.imshow(correlacion_room,
                text_auto=True,
                title="Correlación Satisfaction y Price, Cleanliness Rating, Superhost",
                labels=dict(color="Correlación"), color_continuous_scale= 'Viridis')
fig.update_layout(template="plotly_white")
fig.show()
```

Respecto a Satisfaction de esta matriz podemos sacar en conclusión que el precio no le afecta, la limpieza que tiene una correlación positiva fuerte es lo que más afecta a la satisfacción y Superhost tiene correlación positiva también pero más debil.
En esta matriz también podemos ver varias correlaciones de Superhost bastante intteresantes que estudiaremos más adelante.

### Superhost vs Price, Guest Satisfaction, Cleanliness Rating, City Center
